# Step 5 — Missing values

```python
print("Nulls per column:")
print(df.isna().sum())
print()

# Blank strings or placeholder text in text columns
text_cols = df.select_dtypes(include=["object", "string"]).columns
placeholders = {"", "na", "n/a", "null", "none", "nan", "-", "?"}
for c in text_cols:
    bad = df[c].astype(str).str.strip().str.lower().isin(placeholders).sum()
    print(f"{c:15s} blank/placeholder values: {bad}")
```

```python
# Handling rule (nothing to fill in this file, but this is the pattern to use if values were missing):
#   - identifiers / datetimes / Demand: drop the row (cannot be guessed)
#   - Location: fill with "unknown"
#   - Duration: recompute from the start and end timestamps
df = df.replace(r"^\s*$", np.nan, regex=True)
df = df.dropna(subset=["UserID", "ChargerID", "StartDatetime", "EndDatetime", "Demand"])
df["Location"] = df["Location"].fillna("unknown")
print("Rows after missing-value handling:", len(df))
```

# Step 6 — Duplicates

```python
print("Exact duplicate rows:", df.duplicated().sum())
print("Same charger + same start time:", df.duplicated(["ChargerID", "StartDatetime"]).sum())

# Look at a few duplicates before removing them
df[df.duplicated(keep=False)].sort_values(["ChargerID", "StartDatetime"]).head(6)
```

```python
n = len(df)
df = df.drop_duplicates().reset_index(drop=True)
print("Removed", n - len(df), "exact duplicates; rows left:", len(df))
```

# Step 7 — Fix data types

```python
df["UserID"]    = df["UserID"].astype("string")      # label, not a number to average
df["ChargerID"] = df["ChargerID"].astype("string")
for c in ["ChargerCompany", "ChargerType", "Location"]:
    df[c] = df[c].astype("category")
df["Duration"] = pd.to_numeric(df["Duration"], errors="coerce")
df["Demand"]   = pd.to_numeric(df["Demand"],   errors="coerce")

df.dtypes
```

# Step 8 — Parse the datetime columns

```python
fmt = "%d-%m-%Y %H:%M"
df["StartDatetime"] = pd.to_datetime(df["StartDatetime"], format=fmt, errors="coerce")
df["EndDatetime"]   = pd.to_datetime(df["EndDatetime"],   format=fmt, errors="coerce")

print("Unparseable start:", df["StartDatetime"].isna().sum())
print("Unparseable end:  ", df["EndDatetime"].isna().sum())
print(df[["StartDatetime", "EndDatetime"]].dtypes)
```

```python
start_full = pd.to_datetime(df["StartDay"] + " " + df["StartTime"], format="%d-%m-%Y %H:%M:%S", errors="coerce")
end_full   = pd.to_datetime(df["EndDay"]   + " " + df["EndTime"],   format="%d-%m-%Y %H:%M:%S", errors="coerce")

print("Start mismatch:", (start_full.dt.floor("min") != df["StartDatetime"]).sum())
print("End mismatch:  ", (end_full.dt.floor("min")   != df["EndDatetime"]).sum())

df["StartTimestamp"] = start_full
df["EndTimestamp"]   = end_full
```

```python
start_full = pd.to_datetime(df["StartDay"] + " " + df["StartTime"], format="%d-%m-%Y %H:%M:%S", errors="coerce")
end_full   = pd.to_datetime(df["EndDay"]   + " " + df["EndTime"],   format="%d-%m-%Y %H:%M:%S", errors="coerce")

print("Start mismatch:", (start_full.dt.floor("min") != df["StartDatetime"]).sum())
print("End mismatch:  ", (end_full.dt.floor("min")   != df["EndDatetime"]).sum())

df["StartTimestamp"] = start_full
df["EndTimestamp"]   = end_full
```

# Step 9 — Derive useful time features

```python
df["StartDate"] = df["StartTimestamp"].dt.date
df["Year"]      = df["StartTimestamp"].dt.year
df["Month"]     = df["StartTimestamp"].dt.month
df["MonthName"] = df["StartTimestamp"].dt.month_name()
df["YearMonth"] = df["StartTimestamp"].dt.to_period("M").astype(str)
df["Weekday"]   = df["StartTimestamp"].dt.day_name()
df["DayOfWeek"] = df["StartTimestamp"].dt.dayofweek          # Monday = 0
df["IsWeekend"] = df["DayOfWeek"] >= 5
df["StartHour"] = df["StartTimestamp"].dt.hour
df["EndHour"]   = df["EndTimestamp"].dt.hour
df["CrossesMidnight"] = df["StartTimestamp"].dt.date != df["EndTimestamp"].dt.date

df["TimeOfDay"] = pd.cut(
    df["StartHour"],
    bins=[-1, 5, 11, 17, 21, 23],
    labels=["Night", "Morning", "Afternoon", "Evening", "Late evening"],
)
df[["StartTimestamp", "Year", "Month", "Weekday", "IsWeekend", "StartHour", "TimeOfDay", "CrossesMidnight"]].head()
```

# Step 10 — Elapsed time from timestamps, and check it against Duration

```python
df["ElapsedMin"] = ((df["EndTimestamp"] - df["StartTimestamp"]).dt.total_seconds() / 60).round(2)
df["IdleMin"]    = (df["ElapsedMin"] - df["Duration"]).round(2)

print("End not after start:            ", (df["ElapsedMin"] <= 0).sum())
print("Duration <= 0:                  ", (df["Duration"] <= 0).sum())
print("Duration > elapsed time + 2 min:", (df["Duration"] > df["ElapsedMin"] + 2).sum())
print("Duration > 24 h:                ", (df["Duration"] > 1440).sum())
```

```python
# Remove the invalid rows found above
n = len(df)
valid = (
    (df["ElapsedMin"] > 0)
    & (df["Duration"] > 0)
    & (df["Duration"] <= df["ElapsedMin"] + 2)
    & (df["Duration"] <= 1440)
    & (df["Demand"] > 0)
)
df = df[valid].reset_index(drop=True)
print("Removed", n - len(df), "invalid rows; rows left:", len(df))
```

```python
Overview

The file has one row per EV charging session.
The raw file has 72,856 rows and 13 columns.
Sessions run from 30 Sep 2021 to 30 Sep 2022.
After Phase 1 cleaning there are 72,031 rows, saved as ChargingRecords_cleaned.csv.

Columns

Column	Type	Description
UserID	integer	ID of the EV user. It is an identifier, not a quantity.
ChargerID	integer	ID of the charger used. It is also an identifier.
ChargerCompany	integer (0/1)	Company operating the charger. Company 1 has about 63% of the sessions.
Location	text	Type of site: public area, apartment, resort, hotel, company, public institution, public parking lot, market, restaurant, camping, sightseeing, golf, accommodation, bus garage (14 values).
ChargerType	integer (0/1)	Charger type. About 80% of sessions use type 0.
StartDay, EndDay	text	Session start and end date, dd-mm-YYYY.
StartTime, EndTime	text	Session start and end time, HH:MM:SS. These are the only columns that include seconds.
StartDatetime, EndDatetime	text	Start and end, dd-mm-YYYY HH:MM, at minute precision. They match the Day and Time columns.
Duration	integer	Charging duration in minutes.
Demand	float	Energy demand of the session (0.01 to 97 in the raw file). This is the main variable to analyse.

I inferred these meanings from the column names and values. The source doesn’t say what the 0 and 1 codes mean or what unit Demand uses. It is probably kWh, but I haven’t confirmed that.

Cleaning rules applied (Phase 1)

Dropped 13 exact duplicate rows.
Parsed both datetime columns. Every value parsed, so none were dropped.
Dropped 47 rows where the end time is not after the start time.
Dropped 691 rows with Duration of 0 or less.
Dropped 73 rows where Duration is more than 2 minutes longer than the plugged-in time.
Dropped 1 row with Duration over 24 hours.
No rows had Demand of 0 or less or missing.

Columns added to the cleaned file

ElapsedMin is the plugged-in time in minutes, from the second-level timestamps.
IdleMin is ElapsedMin minus Duration.
FlagIdleOver30 marks 1,445 rows where the car sat idle for more than 30 minutes. These rows were kept.

Known issues and caveats

September 2021 has only 19 sessions, because the data starts on the last day of the month.
155 of the dropped zero-duration rows look like recording errors rather than real zero-length sessions. If you’d rather repair them from the timestamps, that changes the row count.
21 raw rows have a plugged-in time over 24 hours.
Company 1 only operates charger type 0, so comparisons between companies are partly comparisons between charger types.
Demand is right-skewed (skewness about 1.2). I flagged outliers in Phase 2 but did not remove any.
```
